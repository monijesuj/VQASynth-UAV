## 2. 🧠 Load Spatial Reasoning Models

Initialize SpaceOm and SpaceThinker models for advanced spatial reasoning capabilities.


```python
# Initialize spatial reasoning models
class EnhancedSpatialReasoner:
    def __init__(self):
        self.selector = None
        self.models_loaded = False
        self.load_models()
    
    def load_models(self):
        """Load SpaceOm and SpaceThinker models for spatial reasoning"""
        if not HAS_VQASYNTH:
            print("❌ VQASynth-UAV components not available")
            return False
            
        try:
            print("🔄 Loading spatial reasoning models...")
            self.selector = SpatialModelSelector()
            self.selector.load_available_models()
            self.models_loaded = True
            print("✅ Spatial reasoning models loaded successfully!")
            return True
        except Exception as e:
            print(f"❌ Error loading models: {str(e)}")
            return False
    
    def ask_spatial_question(self, images, question, model_choice="SpaceThinker"):
        """Ask spatial reasoning questions about images or point clouds"""
        if not self.models_loaded:
            return "❌ Models not loaded. Please check your setup."
        
        try:
            result = self.selector.ask_spatial_question(
                images=images,
                question=question,
                model_choice=model_choice
            )
            return result
        except Exception as e:
            return f"❌ Error: {str(e)}"
    
    def analyze_point_cloud(self, pcd_path, question):
        """Analyze point cloud with spatial reasoning"""
        if not self.models_loaded:
            return "❌ Models not loaded. Please check your setup."
        
        try:
            result = analyze_point_cloud_with_vlm(
                pcd_path=pcd_path,
                question=question,
                spatial_model_func=self.selector.ask_spatial_question
            )
            return result
        except Exception as e:
            return f"❌ Error analyzing point cloud: {str(e)}"

# Initialize the spatial reasoner
spatial_reasoner = EnhancedSpatialReasoner()

# Test model availability
if spatial_reasoner.models_loaded:
    print("🎯 Available Models:")
    if hasattr(spatial_reasoner.selector, 'models'):
        for model_name in spatial_reasoner.selector.models.keys():
            print(f"  • {model_name}")
    else:
        print("  • SpaceOm (best overall spatial reasoning)")
        print("  • SpaceThinker (most accurate distance estimates)")
else:
    print("⚠️  Models not loaded. Check your model files in ./SpaceOm/ and ./SpaceThinker-Qwen2.5VL-3B/")
```

## 3. 🔮 Point Cloud Processing and Visualization

Convert point clouds to visual representations and analyze 3D spatial relationships.


```python
# Enhanced Point Cloud Processing Class
class PointCloudProcessor:
    def __init__(self):
        self.has_o3d = HAS_O3D
    
    def load_point_cloud(self, pcd_path):
        """Load point cloud from file"""
        if not self.has_o3d:
            print("❌ Open3D not available for point cloud loading")
            return None
        
        try:
            pcd = o3d.io.read_point_cloud(pcd_path)
            if len(pcd.points) == 0:
                print(f"❌ Empty point cloud: {pcd_path}")
                return None
            
            print(f"✅ Loaded point cloud: {len(pcd.points)} points")
            return pcd
        except Exception as e:
            print(f"❌ Error loading point cloud: {str(e)}")
            return None
    
    def analyze_point_cloud_stats(self, pcd):
        """Get basic statistics about the point cloud"""
        if pcd is None:
            return None
        
        points = np.asarray(pcd.points)
        
        stats = {
            'num_points': len(points),
            'x_range': (points[:, 0].min(), points[:, 0].max()),
            'y_range': (points[:, 1].min(), points[:, 1].max()),
            'z_range': (points[:, 2].min(), points[:, 2].max()),
            'centroid': points.mean(axis=0),
            'extent': {
                'x': points[:, 0].max() - points[:, 0].min(),
                'y': points[:, 1].max() - points[:, 1].min(),
                'z': points[:, 2].max() - points[:, 2].min()
            }
        }
        
        return stats
    
    def visualize_point_cloud_3d(self, pcd, title="Point Cloud"):
        """Create interactive 3D visualization of point cloud"""
        if pcd is None:
            return None
        
        points = np.asarray(pcd.points)
        
        # Create 3D scatter plot
        fig = go.Figure(data=[go.Scatter3d(
            x=points[:, 0],
            y=points[:, 1],
            z=points[:, 2],
            mode='markers',
            marker=dict(
                size=2,
                color=points[:, 2],  # Color by height
                colorscale='Viridis',
                opacity=0.6,
                colorbar=dict(title="Height (Z)")
            ),
            name="Point Cloud"
        )])
        
        fig.update_layout(
            title=title,
            scene=dict(
                xaxis_title="X (meters)",
                yaxis_title="Y (meters)",
                zaxis_title="Z (meters)",
                aspectmode='data'
            ),
            height=600
        )
        
        return fig
    
    def convert_to_multiview_images(self, pcd_path, output_size=(512, 512)):
        """Convert point cloud to multiple view images for VLM analysis"""
        try:
            # Use VQASynth-UAV point cloud converter
            view_images = point_cloud_to_image(
                pcd_path=pcd_path,
                output_size=output_size,
                views=['3d', 'front', 'side', 'top']
            )
            print(f"✅ Generated {len(view_images)} view images")
            return view_images
        except Exception as e:
            print(f"❌ Error converting point cloud: {str(e)}")
            return None

# Initialize point cloud processor
pc_processor = PointCloudProcessor()

# Check for existing point cloud files
point_cloud_files = []
if os.path.exists('/home/isr-lab3/James/vqasynth_output/pointclouds'):
    point_cloud_files = [f for f in os.listdir('/home/isr-lab3/James/vqasynth_output/pointclouds') 
                        if f.endswith('.pcd')]
    print(f"🔮 Found {len(point_cloud_files)} point cloud files")
    for i, f in enumerate(point_cloud_files[:3]):  # Show first 3
        print(f"  {i+1}. {f}")
else:
    print("⚠️  No point cloud directory found")

# Also check for the sample point cloud
if os.path.exists('vggt_pointcloud.pcd'):
    print("✅ Found sample point cloud: vggt_pointcloud.pcd")
    point_cloud_files.insert(0, 'vggt_pointcloud.pcd')
```

```python
# Drone Navigation Spatial Analysis
class DroneNavigationAnalyzer:
    def __init__(self, spatial_reasoner):
        self.spatial_reasoner = spatial_reasoner
        self.drone_height = 1.5  # meters
        self.safety_margin = 0.5  # meters
        self.min_clearance = self.drone_height + self.safety_margin
    
    def analyze_navigation_clearance(self, images, question_prefix=""):
        """Analyze navigation clearance for drone flight"""
        questions = [
            f"{question_prefix}What is the minimum clearance height in this scene?",
            f"{question_prefix}Are there any obstacles that would block a {self.drone_height}m tall drone?",
            f"{question_prefix}What is the safest flight path through this area?",
            f"{question_prefix}Which direction offers the most open space for navigation?",
            f"{question_prefix}What is the height of the tallest obstacle that could interfere with flight?"
        ]
        
        results = {}
        for question in questions:
            try:
                answer = self.spatial_reasoner.ask_spatial_question(
                    images=images,
                    question=question,
                    model_choice="SpaceThinker"
                )
                results[question] = answer
            except Exception as e:
                results[question] = f"Error: {str(e)}"
        
        return results
    
    def assess_landing_zones(self, images, question_prefix=""):
        """Assess potential landing zones for drone"""
        questions = [
            f"{question_prefix}Is there a flat, clear area suitable for drone landing?",
            f"{question_prefix}What is the size of the largest open area in this scene?",
            f"{question_prefix}Are there any obstacles that would prevent safe landing?",
            f"{question_prefix}What is the surface area of the best landing zone?",
            f"{question_prefix}How far is the nearest safe landing spot from the current position?"
        ]
        
        results = {}
        for question in questions:
            try:
                answer = self.spatial_reasoner.ask_spatial_question(
                    images=images,
                    question=question,
                    model_choice="SpaceThinker"
                )
                results[question] = answer
            except Exception as e:
                results[question] = f"Error: {str(e)}"
        
        return results
    
    def plan_flight_path(self, images, question_prefix=""):
        """Plan optimal flight path through the scene"""
        questions = [
            f"{question_prefix}What is the optimal flight path to avoid all obstacles?",
            f"{question_prefix}Which route has the most clearance for safe navigation?",
            f"{question_prefix}What is the minimum altitude needed to clear all obstacles?",
            f"{question_prefix}Are there any narrow passages that require careful navigation?",
            f"{question_prefix}What is the maximum safe speed for navigation through this area?"
        ]
        
        results = {}
        for question in questions:
            try:
                answer = self.spatial_reasoner.ask_spatial_question(
                    images=images,
                    question=question,
                    model_choice="SpaceThinker"
                )
                results[question] = answer
            except Exception as e:
                results[question] = f"Error: {str(e)}"
        
        return results

# Initialize drone navigation analyzer
if spatial_reasoner.models_loaded:
    drone_analyzer = DroneNavigationAnalyzer(spatial_reasoner)
    
    print("🚁 Drone Navigation Analysis Capabilities:")
    print("• Navigation clearance assessment")
    print("• Landing zone evaluation") 
    print("• Flight path planning")
    print("• Obstacle avoidance analysis")
    print("• Safety margin calculations")
    
    # Demonstrate with available data
    if sample_view_images is not None:
        print(f"\n🔍 Analyzing navigation clearance...")
        
        # Navigation clearance analysis
        clearance_results = drone_analyzer.analyze_navigation_clearance(
            sample_view_images,
            "For drone navigation, "
        )
        
        print("📊 Navigation Clearance Analysis:")
        for question, answer in list(clearance_results.items())[:2]:  # Show first 2
            print(f"\n🤔 {question}")
            print(f"💡 {answer}")
        
        # Landing zone assessment
        print(f"\n🛬 Assessing landing zones...")
        landing_results = drone_analyzer.assess_landing_zones(
            sample_view_images,
            "For drone landing, "
        )
        
        print("📊 Landing Zone Assessment:")
        for question, answer in list(landing_results.items())[:2]:  # Show first 2
            print(f"\n🤔 {question}")
            print(f"💡 {answer}")
    else:
        print("\n⚠️  No point cloud data available for drone analysis")
        print("💡 Upload point cloud data to test drone navigation capabilities")
else:
    print("❌ Spatial reasoning models not available for drone analysis")
```

## 7. 📊 Advanced Analysis Tools

Spatial measurement and analysis capabilities for comprehensive 3D understanding.


```python
# Advanced Spatial Analysis Tools
class SpatialAnalysisTools:
    def __init__(self, spatial_reasoner):
        self.spatial_reasoner = spatial_reasoner
    
    def measure_distances(self, images, reference_objects):
        """Measure distances between objects in the scene"""
        measurements = {}
        
        for i, obj1 in enumerate(reference_objects):
            for j, obj2 in enumerate(reference_objects[i+1:], i+1):
                question = f"What is the distance between the {obj1} and the {obj2}?"
                
                try:
                    answer = self.spatial_reasoner.ask_spatial_question(
                        images=images,
                        question=question,
                        model_choice="SpaceThinker"
                    )
                    measurements[f"{obj1} to {obj2}"] = answer
                except Exception as e:
                    measurements[f"{obj1} to {obj2}"] = f"Error: {str(e)}"
        
        return measurements
    
    def analyze_heights(self, images, objects):
        """Analyze heights of objects in the scene"""
        height_analysis = {}
        
        for obj in objects:
            question = f"What is the height of the {obj}?"
            
            try:
                answer = self.spatial_reasoner.ask_spatial_question(
                    images=images,
                    question=question,
                    model_choice="SpaceThinker"
                )
                height_analysis[obj] = answer
            except Exception as e:
                height_analysis[obj] = f"Error: {str(e)}"
        
        return height_analysis
    
    def assess_spatial_relationships(self, images, objects):
        """Assess spatial relationships between objects"""
        relationships = {}
        
        relationship_questions = [
            f"Which object is closest to the {objects[0]}?",
            f"What is the relative position of the {objects[1]} compared to the {objects[0]}?",
            f"Are the {objects[0]} and {objects[1]} aligned or offset?",
            f"Which object is higher, the {objects[0]} or the {objects[1]}?",
            f"What is the spatial arrangement of all visible objects?"
        ]
        
        for question in relationship_questions:
            try:
                answer = self.spatial_reasoner.ask_spatial_question(
                    images=images,
                    question=question,
                    model_choice="SpaceOm"
                )
                relationships[question] = answer
            except Exception as e:
                relationships[question] = f"Error: {str(e)}"
        
        return relationships
    
    def generate_spatial_report(self, images, scene_description=""):
        """Generate comprehensive spatial analysis report"""
        report = {
            "scene_description": scene_description,
            "analysis_timestamp": pd.Timestamp.now().strftime("%Y-%m-%d %H:%M:%S"),
            "measurements": {},
            "heights": {},
            "relationships": {},
            "navigation_analysis": {}
        }
        
        # Basic spatial questions
        basic_questions = [
            "What are the main objects visible in this scene?",
            "What is the overall layout and spatial arrangement?",
            "What are the key spatial features for navigation?",
            "What obstacles or constraints are present?"
        ]
        
        for question in basic_questions:
            try:
                answer = self.spatial_reasoner.ask_spatial_question(
                    images=images,
                    question=question,
                    model_choice="SpaceOm"
                )
                report["basic_analysis"][question] = answer
            except Exception as e:
                report["basic_analysis"][question] = f"Error: {str(e)}"
        
        return report

# Initialize advanced analysis tools
if spatial_reasoner.models_loaded:
    analysis_tools = SpatialAnalysisTools(spatial_reasoner)
    
    print("📊 Advanced Spatial Analysis Tools Available:")
    print("• Distance measurements between objects")
    print("• Height analysis of structures")
    print("• Spatial relationship assessment")
    print("• Comprehensive spatial reports")
    print("• Navigation constraint analysis")
    
    # Demonstrate with available data
    if sample_view_images is not None:
        print(f"\n🔍 Running advanced spatial analysis...")
        
        # Example objects for analysis (these would be detected from the scene)
        example_objects = ["building", "tree", "vehicle", "structure"]
        
        # Measure distances
        print("📏 Measuring distances between objects...")
        distance_results = analysis_tools.measure_distances(
            sample_view_images, 
            example_objects[:2]  # Use first 2 objects
        )
        
        print("📊 Distance Measurements:")
        for measurement, result in distance_results.items():
            print(f"  • {measurement}: {result}")
        
        # Analyze heights
        print(f"\n📐 Analyzing object heights...")
        height_results = analysis_tools.analyze_heights(
            sample_view_images,
            example_objects[:2]
        )
        
        print("📊 Height Analysis:")
        for obj, height in height_results.items():
            print(f"  • {obj}: {height}")
    else:
        print("\n⚠️  No point cloud data available for advanced analysis")
        print("💡 Upload point cloud data to test advanced analysis tools")
else:
    print("❌ Spatial reasoning models not available for advanced analysis")
```
